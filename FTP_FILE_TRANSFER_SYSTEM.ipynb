```python
# ============================================
# FTP FILE TRANSFER SYSTEM
# Computer Networks Simulation Application
# ============================================

!pip -q install gradio

import gradio as gr
import os
import shutil

# -------------------------------
# Create FTP Server and Client
# -------------------------------

SERVER_FOLDER = "/content/FTP_Server"
CLIENT_FOLDER = "/content/FTP_Client"

os.makedirs(SERVER_FOLDER, exist_ok=True)
os.makedirs(CLIENT_FOLDER, exist_ok=True)

# Create sample files automatically
with open(os.path.join(SERVER_FOLDER, "sample.txt"), "w") as f:
    f.write("Welcome to FTP File Transfer System!")

with open(os.path.join(SERVER_FOLDER, "network.txt"), "w") as f:
    f.write("Computer Networks - FTP Application")


# -------------------------------
# View Files
# -------------------------------

def view_files():

    files = os.listdir(SERVER_FOLDER)

    if not files:
        return "❌ No files available."

    result = "📂 FTP SERVER FILES\n"
    result += "=" * 45 + "\n\n"

    for i, file in enumerate(files, 1):
        size = os.path.getsize(os.path.join(SERVER_FOLDER, file))
        result += f"{i}. 📄 {file}   ({size} bytes)\n"

    return result


# -------------------------------
# Upload File
# -------------------------------

def upload_file(file):

    if file is None:
        return "❌ Please select a file."

    filename = os.path.basename(file.name)

    destination = os.path.join(
        SERVER_FOLDER,
        filename
    )

    shutil.copy(file.name, destination)

    return f"""
✅ FILE UPLOADED SUCCESSFULLY

File Name : {filename}
Server    : FTP Server
Status    : Upload Completed
"""


# -------------------------------
# Download File
# -------------------------------

def download_file(filename):

    filename = filename.strip()

    source = os.path.join(
        SERVER_FOLDER,
        filename
    )

    if not os.path.exists(source):
        return None, f"❌ File '{filename}' not found."

    destination = os.path.join(
        CLIENT_FOLDER,
        filename
    )

    shutil.copy(source, destination)

    return destination, f"""
✅ FILE DOWNLOADED SUCCESSFULLY

File Name  : {filename}
Destination: FTP Client
Status     : Download Completed
"""


# -------------------------------
# Delete File
# -------------------------------

def delete_file(filename):

    filename = filename.strip()

    path = os.path.join(
        SERVER_FOLDER,
        filename
    )

    if not os.path.exists(path):
        return f"❌ File '{filename}' not found."

    os.remove(path)

    return f"""
🗑️ FILE DELETED SUCCESSFULLY

File Name : {filename}
Status    : Removed from FTP Server
"""


# ============================================
# APPLICATION UI
# ============================================

with gr.Blocks(
    title="FTP File Transfer System"
) as app:

    gr.Markdown(
        """
        # 🌐 FTP FILE TRANSFER SYSTEM
        ### Computer Networks Simulation Application

        **File Transfer Protocol (FTP) - Client / Server Model**
        """
    )

    gr.Markdown(
        """
        This application demonstrates file transfer between
        an FTP Client and FTP Server.
        """
    )

    # --------------------------------
    # SERVER FILE LIST
    # --------------------------------

    gr.Markdown("## 📂 FTP Server")

    view_button = gr.Button(
        "🔄 View Available Files"
    )

    files_output = gr.Textbox(
        label="Server Files",
        value=view_files(),
        lines=6
    )

    view_button.click(
        view_files,
        outputs=files_output
    )

    # --------------------------------
    # UPLOAD
    # --------------------------------

    gr.Markdown("## 📤 Upload File")

    upload_input = gr.File(
        label="Select File"
    )

    upload_button = gr.Button(
        "📤 Upload to FTP Server"
    )

    upload_output = gr.Textbox(
        label="Upload Output",
        lines=5
    )

    upload_button.click(
        upload_file,
        inputs=upload_input,
        outputs=upload_output
    ).then(
        view_files,
        outputs=files_output
    )

    # --------------------------------
    # DOWNLOAD
    # --------------------------------

    gr.Markdown("## 📥 Download File")

    download_input = gr.Textbox(
        label="Enter File Name",
        value="sample.txt"
    )

    download_button = gr.Button(
        "📥 Download from FTP Server"
    )

    download_output = gr.File(
        label="Downloaded File"
    )

    download_status = gr.Textbox(
        label="Download Output",
        lines=5
    )

    download_button.click(
        download_file,
        inputs=download_input,
        outputs=[
            download_output,
            download_status
        ]
    )

    # --------------------------------
    # DELETE
    # --------------------------------

    gr.Markdown("## 🗑️ Delete File")

    delete_input = gr.Textbox(
        label="Enter File Name",
        value="network.txt"
    )

    delete_button = gr.Button(
        "🗑️ Delete File"
    )

    delete_output = gr.Textbox(
        label="Delete Output",
        lines=5
    )

    delete_button.click(
        delete_file,
        inputs=delete_input,
        outputs=delete_output
    ).then(
        view_files,
        outputs=files_output
    )


# ============================================
# RUN APPLICATION
# ============================================

app.launch(share=True)
```
